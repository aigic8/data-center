- How many nodes are GPU based in Alibaba data?
  
  There are `1213` nodes with GPU out of `1523` nodes. Which means about `80%` of the nodes have at least one GPU. Which can give us a good estimate on how many nodes in a GPU cluster are GPU based.

- How many GPUs does each node have on average?

  These are the number of GPU used in the nodes:
  - `1: 024, 02.00%`
  - `2: 518, 42.70%`
  - `4: 054, 04.45%`
  - `8: 617, 50.87%`
  
  There two things we can understand from this analysis:
  - All the GPUs are even numbers (Why?)
  - Most (more than 92%) of the of the nodes **either have 2 or 8 gpus**

```python
# number of gpus of each node
total_gpu_nodes = 1213
for i in range(1, 9):
  nodes_with_i_gpus = nodes_df.iloc[nodes_df["gpu"].to_numpy() == i]
  nodes_len = len(nodes_with_i_gpus)
  if nodes_len > 0:
    print(f'{i}: {len(nodes_with_i_gpus)}, {nodes_len*100.0/total_gpu_nodes}%')
```
