```python
from pathlib import Path

import polars as pl
from icecream import ic
```

```python

df_spec = pl.read_csv("../data/alibaba-2020/pai_machine_spec.csv")
df_spec.head()
```

```python
df_spec['gpu_type'].unique()
```

```python
df_metric = pl.read_csv('../data/alibaba-2020/pai_machine_metric.csv')
df_metric = df_metric[['worker_name', 'machine', 'start_time', 'end_time', 'machine_gpu', 'machine_cpu']]
df_metric.head()
```

```python
start_time = df_metric['start_time'].min()
end_time = df_metric['end_time'].max()
ic(start_time)
ic(end_time)
ic((end_time - start_time)/3600);
```

```python
df_merged = df_metric.join(df_spec[['gpu_type', 'cap_gpu', 'cap_cpu', 'machine']], on='machine', how='left')
df_merged.head()

parent_path = '../data/alibaba-2020/generated'
Path(parent_path).mkdir(parents=True, exist_ok=True)
df_merged.write_csv(f'{parent_path}/merged_dataframes.csv')
```
