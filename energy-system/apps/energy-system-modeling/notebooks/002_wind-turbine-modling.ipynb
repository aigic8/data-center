```python
df = pd.read_csv(DATA_PATH, parse_dates=["time"], index_col="time")
df.head()
```

```python
t = wt.WindTurbine(
    cut_in_speed=2.0,
    cut_out_speed=9.0,
    rated_speed=9,
    efficiency=0.95,
    height=20,
    rated_power=2,
)
```

```python
# Change the Height ######################
ws_col_name = f"WS{t.height}"
df[ws_col_name] = wt.wind_change_height(df["WS10m"], ref_height=10, hub_height=t.height)
df.head()
```

```python
# Wind Turbine Model #####################
df["power"] = wt.wt_power(df[ws_col_name], t)
df.head()
```
